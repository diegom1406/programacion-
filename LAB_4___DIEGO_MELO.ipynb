```python
# ==========================================
# 1. PREPARACIÓN DEL ENTORNO
# ==========================================

def mostrar_resultado(titulo, contenido):
    print("=" * 55)
    print(titulo)
    print("=" * 55)
    print(contenido)
    print()

print("Sistema experto iniciado correctamente.")
```

```python
# ==========================================
# 2. HECHOS DEL USUARIO
# ==========================================

hechos = {
    "accion": False,
    "adrenalina": False,
    "familia": False,
    "ciencia_ficcion": True,
    "comedia": False,
    "poco_tiempo": False
}

print("HECHOS DEL USUARIO:")
for hecho, valor in hechos.items():
    print(f"- {hecho}: {valor}")
```

```python
# ==========================================
# 3. BASE DE CONOCIMIENTO
# ==========================================

reglas = [
    {
        "nombre": "R1",
        "condiciones": ["accion", "adrenalina"],
        "resultado": "Acción"
    },
    {
        "nombre": "R2",
        "condiciones": ["familia"],
        "resultado": "Familiar"
    },
    {
        "nombre": "R3",
        "condiciones": ["ciencia_ficcion"],
        "resultado": "Ciencia ficción"
    },
    {
        "nombre": "R4",
        "condiciones": ["comedia", "familia"],
        "resultado": "Comedia familiar"
    },
    {
        "nombre": "R5",
        "condiciones": ["poco_tiempo"],
        "resultado": "Película corta"
    }
]

print("BASE DE CONOCIMIENTO:")
for regla in reglas:
    print(f"{regla['nombre']}: SI {regla['condiciones']} → ENTONCES {regla['resultado']}")
```

```python
# ==========================================
# 4. MOTOR DE INFERENCIA
# ==========================================

def motor_inferencia(hechos, reglas):
    conclusiones = []
    reglas_activadas = []

    # Revisamos una regla a la vez
    for regla in reglas:

        # all() es True solamente cuando TODAS
        # las condiciones de la regla se cumplen.
        cumple = all(
            hechos.get(condicion, False)
            for condicion in regla["condiciones"]
        )

        # Si la regla se cumple, la activamos.
        if cumple:
            reglas_activadas.append(regla["nombre"])
            conclusiones.append(regla["resultado"])

    return conclusiones, reglas_activadas


conclusiones, reglas_activadas = motor_inferencia(hechos, reglas)

print("REGLAS ACTIVADAS:", reglas_activadas)
print("CONCLUSIONES GENERADAS:", conclusiones)
```

```python
# ==========================================
# 5. EXPLICACIÓN DE LA DECISIÓN
# ==========================================

def explicar_decision(hechos, reglas, reglas_activadas):

    print("=" * 55)
    print("EXPLICACIÓN DEL SISTEMA")
    print("=" * 55)

    if not reglas_activadas:
        print("No se activó ninguna regla.")
        print("El sistema no tiene suficiente conocimiento para recomendar.")
        return

    for regla in reglas:

        if regla["nombre"] in reglas_activadas:

            print(f"\n✓ Se activó {regla['nombre']}")

            for condicion in regla["condiciones"]:
                print(f"  - {condicion}: {hechos.get(condicion, False)}")

            print(f"  → Recomendación: {regla['resultado']}")


explicar_decision(hechos, reglas, reglas_activadas)
```

```python
# ==========================================
# 6. PRUEBA DE ESCENARIOS
# ==========================================

escenarios = {
    "Escenario 1 - Familiar": {
        "accion": False, "adrenalina": False, "familia": True,
        "ciencia_ficcion": False, "comedia": False, "poco_tiempo": False
    },
    "Escenario 2 - Ciencia ficción": {
        "accion": False, "adrenalina": False, "familia": False,
        "ciencia_ficcion": True, "comedia": False, "poco_tiempo": False
    },
    "Escenario 3 - Poco tiempo": {
        "accion": False, "adrenalina": False, "familia": False,
        "ciencia_ficcion": False, "comedia": False, "poco_tiempo": True
    }
}

for nombre, hechos_escenario in escenarios.items():

    conclusiones, activadas = motor_inferencia(
        hechos_escenario, reglas
    )

    print("\n" + "=" * 55)
    print(nombre)
    print("=" * 55)
    print("Reglas activadas:", activadas)
    print("Recomendación:", conclusiones if conclusiones else "Sin recomendación")
```
